```python
import numpy as np
import matplotlib.pyplot as plt

import matplotlib as mpl
import matplotlib.colors as mcolors

%matplotlib widget
```

```python
cmap_choice = 'rainbow_r'
n_files = 11

cmap = mpl.colormaps[cmap_choice]

# Optional: To adjust the number of color steps (resample)
cmap_resampled = mpl.colormaps[cmap_choice].resampled(n_files)
```

```python
cmap_resampled
```

```python

rgb_hex_strings = [mpl.colors.to_hex(cmap_resampled(i)) for i in range(n_files)]
```
